```python
from nltk.stem import PorterStemmer , LancasterStemmer , SnowballStemmer , RegexpStemmer
```

```python
stemming=PorterStemmer()
```

```python
words = [
    "running",
    "flies",
    "happily",
    "denied",
    "studying",
    "connections",
    "beauties",
    "troubling",
    "maximums",
    "activation"
]
```

```python
for i in words:
    print(i , "---------->",stemming.stem(i))
```

```python
lancaster =  LancasterStemmer()
```

```python
for i in words:
    print(i , "---------->",lancaster.stem(i))
```

```python
snowball = SnowballStemmer(language = "english")
```

```python
for i in words:
    print(i , "---------->",snowball.stem(i))
```

```python
regex = RegexpStemmer("ing$|edly$|ed$|es$|s$|ion$|ly$")
```

```python
for i in words:
    print(i , "---------->",regex.stem(i))
```

**Lemmatizer**

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmitizer = WordNetLemmatizer()
```

```python
for i in words:
    print(i , "---------->",lemmitizer.lemmatize(i))
```

```python
for i in words:
    print(i , "---------->",lemmitizer.lemmatize(i,pos="v"))
```
