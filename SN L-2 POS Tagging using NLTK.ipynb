```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag
```

```python
# Download required resources
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
```

```python
# Input text
text = "This is a practical example of POS tagging."
```

```python
# Tokenization
words = word_tokenize(text)
```

```python
# Convert words to lowercase
lower_words = [word.lower() for word in words]
```

```python
# POS Tagging
pos_result = pos_tag(lower_words)
```

```python
# Display result
print("Original Text:")
print(text)

print("\nTokens:")
print(lower_words)

print("\nPOS Tags:")
print(pos_result)
```
